## 1. Tasa de conversión de una tienda (input, int, división por cero)

```python
shop_name = input("Name of your shop: ")
shop_visits = int(input("Number of visits: "))
shop_sales = int(input("Number of sales: "))
if shop_visits > 0:
    conversion_rate = shop_sales/shop_visits
    result = f"{conversion_rate:.2%}"
else:
    result = "Cannot be calculated"

print(f"Shop: {shop_name}, Visits: {shop_visits}, Sales: {shop_sales}, Conversion: {result}")
```

## 2. Indicadores de campaña: CTR, CPC y ROI (float, if-elif-else)

```python
campaña = input("Nombre de campaña: ")
impresiones = int(input("Numero de impresiones: "))
clics = int(input("Numero de clics: "))
inversion = float(input("Inversion en EUR: " ))
ingresos = float(input("Ingresos en EUR: "))


ctr = (clics / impresiones) * 100
cpc = inversion / clics
beneficio = ingresos - inversion
roi = (beneficio / inversion) *100


print(f"Campaña: {campaña}")
print(f"CTR: {ctr:.2f}%")
print(f"CPC: {cpc:.2f}€")
print(f"Beneficio: {beneficio:.2f}€")
print(f"ROI: {roi:.2f}%")

if roi > 0:
    print("La campaña es rentable")
elif roi == 0:
    print("La campaña no gana ni pierde")
else:
    print("La campaña no es rentable")
```

## 6. Análisis de ventas: total, media, máximo y mínimo (sum, len, max, min)

```python
ventas = [1200, 950, 1430, 1100, 1600]
total = sum(ventas)
media = total / len(ventas)


print(f"Total: {total}€")
print(f"Media: {media:.2f}€")
print(f"Maxima: {max(ventas)}€ Minima: {min(ventas)}€")

ventas.append(1750)
total = sum(ventas)
media = total / len(ventas)
print(f"Nuevo total: {total}€ Nueva media: {media:.2f}€")

```

## 7. Comparativa A/B: suma y media de dos grupos (if-elif-else)

```python
grupo_a = [10, 15, 10, 20]
grupo_b = [15, 25, 20, 30]
total_grupo_a = sum(grupo_a)
total_grupo_b = sum(grupo_b)

media_grupo_a = total_grupo_a / len(grupo_a)
media_grupo_b = total_grupo_b / len(grupo_b)

print(f"Grupo A → total: {total_grupo_a}, media: {media_grupo_a:.2f}")
print(f"Grupo B → total: {total_grupo_b}, media: {media_grupo_b:.2f}")


if media_grupo_a > media_grupo_b:
    print("Grupo A es el mejor.")
elif media_grupo_a < media_grupo_b:
    print("Grupo B es el mejor.")
else:
    print("Los dos grupos son iguales.")
```

## 9. Precio de entrada de cine según la edad (if-elif-else, try-except)

```python
try:
    edad = int(input("Introduce tu edad: "))

    if edad < 0:
        print("Edad no válida: no puede ser negativa")
    else:
        if edad < 5:
            tipo = "Gratis"
            precio = 0
        elif edad < 18:
            tipo = "Descuento de estudiante"
            precio = 5
        elif edad < 65:
            tipo = "Entrada general"
            precio = 17
        else:
            tipo = "Descuento de adulto mayor"
            precio = 4

        print(f"Edad: {edad} años → {tipo}: {precio}€")

except ValueError:
    print("Error: introduce un número entero")
```

## 11. Evaluador de notas: número a letra A-F (validación de rango, try-except)

```python
try:
    nota = int(input("Introduce la nota (0-100): "))

    if nota < 0 or nota > 100:
        print("Error: La nota debe estar entre 0 y 100")
    else:
        if nota >= 90:
            letra = "A (Excelente)"
        elif nota >= 80:
            letra = "B (Bueno)"
        elif nota >= 70:
            letra = "C (Suficiente)"
        else:
            letra = "F (Reprobado)"

        print(f"Nota: {nota} → {letra}")

except ValueError:
    print("Error: introduce un número entero")
```

## 12. Calculadora con funciones (def, return, división por cero)

```python
def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        return None
    return a / b

try:
    n1 = int(input("Inserta el número 1: "))
    n2 = int(input("Inserta el número 2: "))

    print(f"{n1} + {n2} = {sumar(n1, n2)}")
    print(f"{n1} - {n2} = {restar(n1, n2)}")
    print(f"{n1} * {n2} = {multiplicar(n1, n2)}")

    resultado = dividir(n1, n2)
    if resultado is None:
        print("No se puede dividir por 0")
    else:
        print(f"{n1} / {n2} = {resultado:.2f}")

except ValueError:
    print("Error: introduce números enteros")
```

## 13. Estadísticas de ventas con funciones (parámetro lista, return)

```python
ventas = [120, 80, 150, 200]

def calcular_total(lista):
    result = sum(lista)
    return result

def calcular_media(lista):
    result = calcular_total(lista) / len(lista)
    return result

def obtener_maximo(lista):
    result = max(lista)
    return result

total = calcular_total(ventas)
media = calcular_media(ventas)
maximo = obtener_maximo(ventas)

print(f"Total de ventas: {total}€")
print(f"Media de ventas: {media:.2f}€")
print(f"Venta más alta: {maximo}€")
```

## 14. Hamburguesería: ventas, stock y recarga (while dentro de una función)


```python
def gestionar_ventas(inventario):
    print(f"El inventario contiene {inventario} hamburguesas")

    while inventario > 0:
        if inventario <= 10:
            print("Advertencia: El inventario está casi vacío.")

        num_hamburguesas = int(input("¿Cuántas hamburguesas quiere el cliente? "))

        if num_hamburguesas <= 0:
            print("Error: el pedido debe ser mayor que 0.")
        elif num_hamburguesas > inventario:
            print(f"No hay suficientes hamburguesas. Solo quedan {inventario}.")
        else:
            inventario -= num_hamburguesas
            print(f"El cliente ha pedido {num_hamburguesas} hamburguesas. El inventario ahora tiene {inventario} hamburguesas.")

        if inventario == 0:
            print("Inventario agotado.")
            respuesta = input("¿Quieres hacer una recarga? (si/no): ").strip().lower()
            if respuesta == "si":
                cantidad = int(input("¿Cuántas hamburguesas quieres añadir? "))
                if cantidad > 0:
                    inventario += cantidad
                    print(f"Recarga hecha. El inventario ahora tiene {inventario} hamburguesas.")
                else:
                    print("Error: la recarga debe ser mayor que 0.")

    print("No hay recarga. Cerramos la hamburguesería.")

gestionar_ventas(100)
```
