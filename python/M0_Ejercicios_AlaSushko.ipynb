### 🔴 Ejercicio 1.5 — Normalización de SKU y Cálculo de Margen Neto *(Reto Integrador)*
**Contexto:** El departamento financiero necesita calcular el margen real de un producto, pero el catálogo mezcla códigos en minúsculas y precios con IVA incluido (21 %) en formato texto.

**Tu tarea:**
1. Normaliza `sku_raw` quitando espacios laterales y pasándolo todo a **MAYÚSCULAS** (`sku_limpio`).
2. Limpia `pvp_con_iva_raw` y conviértelo a `float`.
3. Calcula el `pvp_sin_iva` dividiendo el precio con IVA entre `1.21` (redondeado a 2 decimales).
4. Convierte `coste_proveedor_raw` a `float` y calcula:
   * `margen_bruto`: `pvp_sin_iva - coste_proveedor` (redondeado a 2 decimales).
   * `margen_pct`: `(margen_bruto / pvp_sin_iva) * 100` (redondeado a 1 decimal).
5. Imprime la ficha financiera completa del producto.

> ✅ **Resultado esperado:** `SKU: ELEC-2026-TV55 | PVP sin IVA: 500.0€ | Margen: 179.5€ (35.9%)`

````python
# Datos crudos de catálogo y finanzas
sku_raw = "  elec-2026-tv55  "
pvp_con_iva_raw = "605,00€"
coste_proveedor_raw = "320.50"

# Escribe aquí tu solución para el Ejercicio 1.5

# 1. SKU normalizado en mayúsculas
sku_limpio = sku_raw.strip().upper()

# 2. PVP con IVA: quitar €, espacios y cambiar coma decimal por punto
pvp_en_proceso = pvp_con_iva_raw.replace("€", "").strip()
pvp_en_proceso = pvp_en_proceso.replace(",", ".")
pvp_con_iva = float(pvp_en_proceso)

# 3. Quitar el IVA del 21 %
pvp_sin_iva = round(pvp_con_iva / 1.21, 2)

# 4. Coste y margen
coste_proveedor = float(coste_proveedor_raw)
margen_bruto = round(pvp_sin_iva - coste_proveedor, 2)
margen_pct = round((margen_bruto / pvp_sin_iva) * 100, 1)

# 5. Ficha financiera
print(f"SKU: {sku_limpio} | PVP sin IVA: {pvp_sin_iva}€ | Margen: {margen_bruto}€ ({margen_pct}%)")
````

### 🔴 Ejercicio 2.5 — Agrupación y Resumen por Grupo con Diccionarios *(Reto Integrador)*
**Contexto:** En analítica de datos necesitamos agrupar filas por una categoría (como haría un `GROUP BY` en SQL) para comparar métricas entre grupos. Vas a resolver el análisis completo del Test A/B usando **únicamente diccionarios y listas nativas de Python, sin importar ninguna librería**.

**Tu tarea:**
1. Crea un diccionario acumulador llamado `resumen_grupos` con la estructura inicial:
   ```python
   resumen_grupos = {
       "A": {"usuarios": 0, "convertidos": 0, "tiempos": []},
       "B": {"usuarios": 0, "convertidos": 0, "tiempos": []},
   }
   ```
2. Recorre la lista `experimento_ab` con un bucle `for`:
   * Obtén el grupo de la fila (`grupo = fila["grupo"]`).
   * Suma `1` al contador `resumen_grupos[grupo]["usuarios"]`.
   * Si `fila["convirtio"]` es `True`, suma `1` a `resumen_grupos[grupo]["convertidos"]`.
   * Si `fila["tiempo_seg"]` **no es `None`**, añade ese tiempo a la lista `resumen_grupos[grupo]["tiempos"]`.
3. Crea un diccionario final `resultado_ab = {}` recorriendo `resumen_grupos.items()` para calcular en cada grupo:
   * `tasa_conv_pct`: `(convertidos / usuarios) * 100`, redondeada a 1 decimal.
   * `tiempo_medio_seg`: promedio de la lista `tiempos`, redondeado a 1 decimal.
4. Imprime el resumen comparativo del Grupo A frente al Grupo B.

> 💡 **Pista:** Accede a los valores internos encadenando corchetes, por ejemplo: `resumen_grupos[grupo]["usuarios"] += 1`.
> ✅ **Resultado esperado:** `Grupo A: 0.0% conversión (40.0s) | Grupo B: 100.0% conversión (102.5s)`

````python
# Usamos la misma tabla experimento_ab (sin importar librerías externas)
experimento_ab = [
    {"user": "U1", "grupo": "A", "tiempo_seg": 40, "convirtio": False},
    {"user": "U2", "grupo": "B", "tiempo_seg": 95, "convirtio": True},
    {"user": "U3", "grupo": "A", "tiempo_seg": None, "convirtio": False},
    {"user": "U4", "grupo": "B", "tiempo_seg": 110, "convirtio": True},
]

# Escribe aquí tu solución para el Ejercicio 2.5 usando diccionarios

# 1. Diccionario acumulador por grupo
resumen_grupos = {
    "A": {"usuarios": 0, "convertidos": 0, "tiempos": []},
    "B": {"usuarios": 0, "convertidos": 0, "tiempos": []},
}

# 2. Recorrer las filas y acumular en su grupo
for usuario in experimento_ab:
    grupo = usuario["grupo"]
    resumen_grupos[grupo]["usuarios"] += 1
    if usuario["convirtio"]:
        resumen_grupos[grupo]["convertidos"] += 1
    if usuario["tiempo_seg"] is not None:
        resumen_grupos[grupo]["tiempos"].append(usuario["tiempo_seg"])

# 3. Calcular las métricas finales de cada grupo
resultado_ab = {}
for grupo, datos in resumen_grupos.items():
    tasa_conv_pct = round((datos["convertidos"] / datos["usuarios"]) * 100, 1)
    tiempo_medio_seg = round(sum(datos["tiempos"]) / len(datos["tiempos"]), 1)
    resultado_ab[grupo] = {"tasa_conv_pct": tasa_conv_pct, "tiempo_medio_seg": tiempo_medio_seg}

# 4. Resumen comparativo
a = resultado_ab["A"]
b = resultado_ab["B"]
print(f"Grupo A: {a['tasa_conv_pct']}% conversión ({a['tiempo_medio_seg']}s) | Grupo B: {b['tasa_conv_pct']}% conversión ({b['tiempo_medio_seg']}s)")
````

### 🟢 Ejercicio 4.1 — Primera Función Financiera con `return` *(Calentamiento)*
**Contexto:** Necesitas calcular el precio final con IVA de cientos de productos. En lugar de repetir la fórmula, crearás una función reutilizable con un valor por defecto.

**Tu tarea:**
1. Define una función llamada `calcular_iva(precio_base, tipo_iva=0.21)` que devuelva (`return`) `round(precio_base * (1 + tipo_iva), 2)`.
2. Llama a tu función con `100` (usará el IVA general del 21 % por defecto) y guarda el resultado en `precio_general`.
3. Llama a tu función con `100` y `tipo_iva=0.10` (IVA reducido del 10 %) y guarda el resultado en `precio_reducido`.
4. Imprime ambos resultados.

> ⚠️ **Recuerda:** Usa `return` dentro de la función, no `print()`. Así podrás guardar el resultado en variables.
> ✅ **Resultado esperado:** `General: 121.0€ | Reducido: 110.0€`

````python
# Escribe aquí tu solución para el Ejercicio 4.1

# 1. Definir la función con un valor por defecto
def calcular_iva(precio_base, tipo_iva=0.21):
    return round(precio_base * (1 + tipo_iva), 2)

# 2. IVA general: no indicamos tipo_iva, se usa 0.21
precio_general = calcular_iva(100)

# 3. IVA reducido: indicamos tipo_iva=0.10
precio_reducido = calcular_iva(100, tipo_iva=0.10)

# 4. Resultado
print(f"General: {precio_general}€ | Reducido: {precio_reducido}€")
````

   * En caso contrario, convierta `texto` a `str`, elimine espacios laterales y devuelva el texto en formato Título (`.strip().title()`).

### 🟠 Ejercicio 4.4 — Etapa `Transform` Modular *(Intermedio-Alto)*
**Contexto:** Ahora vas a reutilizar las funciones que creaste en los ejercicios **4.2** y **4.3** dentro de una nueva función llamada `transformar_datos(filas_crudas)`. ¡Esto es programación modular real!

**Tu tarea:**
1. Define `transformar_datos(filas_crudas)`.
2. Dentro de ella, crea una lista `datos_limpios = []` y recorre `filas_crudas`:
   * Usa `es_registro_valido(fila)` para filtrar las filas corruptas.
   * Si es válida, añade a `datos_limpios` un nuevo diccionario con:
     * `"categoria"`: usando `normalizar_texto(fila.get("categoria"))`
     * `"ingreso"`: `round(fila["unidades"] * fila["precio"], 2)`
3. Devuelve `datos_limpios` con `return`.
4. Ejecuta la función sobre `dataset_ventas_raw` e imprime el resultado.

> ✅ **Resultado esperado:** `[{'categoria': 'Tech', 'ingreso': 200.0}, {'categoria': 'Tech', 'ingreso': 150.0}]`

````python
# Dataset crudo para probar la transformación
dataset_ventas_raw = [
    {"categoria": " tech ", "unidades": 2, "precio": 100.0},
    {"categoria": "hogar", "unidades": -1, "precio": 50.0},
    {"categoria": "TECH", "unidades": 1, "precio": 150.0},
    {"categoria": "moda", "unidades": 3, "precio": None},
]

# Escribe aquí tu solución para el Ejercicio 4.4

# 1. Función de transformación que reutiliza las funciones de 4.2 y 4.3
def transformar_datos(filas_crudas):
  # Tabla nueva vacía donde guardaremos las filas limpias
    datos_limpios = []
    # Recorrer la tabla cruda fila a fila
    for fila in filas_crudas:
       # Filtro: solo seguimos si la fila es válida (función de 4.3)
        if es_registro_valido(fila):
             # Añadir una fila nueva y limpia a la tabla
            datos_limpios.append({
                "categoria": normalizar_texto(fila.get("categoria")),
                "ingreso": round(fila["unidades"] * fila["precio"], 2),
            })

    return datos_limpios

# 2. Ejecutar la función sobre el dataset
print(transformar_datos(dataset_ventas_raw))
````

### 🔴 Ejercicio 4.5 — Pipeline ETL Completo y Reporte Ejecutivo *(Reto Integrador Final)*
**Contexto:** Has llegado al reto final del módulo. Vas a construir la función orquestadora `generar_reporte_etl(filas_crudas)` que conecta la limpieza con el cálculo de KPIs ejecutivos y métricas de calidad del dato.

**Tu tarea:**
1. Define `generar_reporte_etl(filas_crudas)`.
2. Llama dentro a `limpios = transformar_datos(filas_crudas)` (tu función del Ejercicio 4.4).
3. Calcula:
   * `registros_entrada`: `len(filas_crudas)`
   * `registros_validos`: `len(limpios)`
   * `tasa_calidad_pct`: `(registros_validos / registros_entrada) * 100` redondeado a 1 decimal (protegido con `if registros_entrada > 0 else 0.0`).
   * `facturacion_total`: suma de `"ingreso"` en `limpios`, redondeada a 2 decimales.
   * `facturacion_por_categoria`: diccionario con el total acumulado por cada categoría limpia.
4. Devuelve (`return`) un diccionario con todas esas métricas ejecutivas e imprímelo de forma clara.

> ✅ **Resultado esperado:**
> `{'registros_entrada': 4, 'registros_validos': 2, 'tasa_calidad_pct': 50.0, 'facturacion_total': 350.0, 'facturacion_por_categoria': {'Tech': 350.0}}`

````python
# Escribe aquí tu solución para el Ejercicio 4.5 (Reto Integrador Final)

# 1. Función orquestadora del pipeline
def generar_reporte_etl(filas_crudas):
    # 2. Transform: reutilizamos la función del 4.4
    limpios = transformar_datos(filas_crudas)

    # 3. Métricas de calidad del dato
    registros_entrada = len(filas_crudas)
    registros_validos = len(limpios)
    tasa_calidad_pct = round((registros_validos / registros_entrada) * 100, 1) if registros_entrada > 0 else 0.0
    #if registros_entrada > 0:
        #tasa_calidad_pct = round((registros_validos / registros_entrada) * 100, 1)
    #else:
        #tasa_calidad_pct = 0.0

    # KPI de negocio: facturación total
    facturacion_total = 0
    for fila in limpios:
        facturacion_total += fila["ingreso"]
    facturacion_total = round(facturacion_total, 2)
    #facturacion_total = round(sum(fila["ingreso"] for fila in limpios), 2)

    # Facturación por categoría (agrupación, como en el 3.5)
    facturacion_por_categoria = {}
    for fila in limpios:
        categoria = fila["categoria"]
        facturacion_por_categoria[categoria] = facturacion_por_categoria.get(categoria, 0.0) + fila["ingreso"]


    # 4. Devolver todas las métricas en un diccionario
    return {
        "registros_entrada": registros_entrada,
        "registros_validos": registros_validos,
        "tasa_calidad_pct": tasa_calidad_pct,
        "facturacion_total": facturacion_total,
        "facturacion_por_categoria": facturacion_por_categoria
    }

# Ejecutar el pipeline y mostrar el reporte
reporte = generar_reporte_etl(dataset_ventas_raw)
print(reporte)
````
