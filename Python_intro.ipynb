## 1. Tasa de conversión

```python
shop_name = input("Name of your shop: ")
shop_visits = int(input("Number of visits: "))
shop_sales = int(input("Number of sales: "))
if shop_visits > 0:
    conversion_rate = shop_sales/shop_visits
    result = f"{conversion_rate:.2%}"
else:
    result = "Cannot be calculated"

print(f"Shop: {shop_name}, Visits: {shop_visits}, Sales: {shop_sales}, Conversion: {result}")
```

## 2. Calculadora de indicadores (CTR, CPC, ROI)

```python
campana = input("Nombre de campana: ")
impresiones = int(input("Numero de impresiones: "))
clics = int(input("Numero de clics: "))
inversion = float(input("Inversion en EUR: " ))
ingresos = float(input("Ingresos en EUR: "))


ctr = (clics / impresiones) * 100
cpc = inversion / clics
beneficio = ingresos - inversion
roi = (beneficio / inversion) *100


print(f"Campana: {campana}")
print(f"CTR: {ctr:.2f}%")
print(f"CPC: {cpc:.2f}€")
print(f"Beneficio: {beneficio:.2f}€")
print(f"ROI: {roi:.2f}%")

if roi > 0:
    print("La campana es rentable")
elif roi == 0:
    print("La campana no gana ni pierde")
else:
    print("La campana no es rentable")
```

## 3. Campaña de marketing: tipos de datos

```python
campaign = input("Enter campaign name: ")
impressiones = int(input("Number of impressions: "))
clicks = int(input("Number of clicks: "))
ad_spend = float(input("Enter ad spend in EUR: " ))
campaign_active = input("Is the campaign active? (yes/no): ").strip().lower() == "yes"


print("Campaign:", campaign, type(campaign))
print("Impressions:", impressiones, type(impressiones))
print("Clicks:", clicks, type(clicks))
print("Ad spend:", ad_spend, type(ad_spend))
print("Active:", campaign_active, type(campaign_active))
```

## 6. Analizar ventas

```python
ventas = [1200, 950, 1430, 1100, 1600]
total = sum(ventas)
media = total / len(ventas)


print(f"Total: {total}€")
print(f"Media: {media:.2f}€")
print(f"Maxima: {max(ventas)}€ Minima: {min(ventas)}€")

ventas.append(1750)
total = sum(ventas)
media = total / len(ventas)
print(f"Nuevo total: {total}€ Nueva media: {media:.2f}€")

```

## 7. Comparativa A/B

```python
grupo_a = [10, 15, 10, 20]
grupo_b = [15, 25, 20, 30]
total_grupo_a = sum(grupo_a)
total_grupo_b = sum(grupo_b)

media_grupo_a = total_grupo_a / len(grupo_a)
media_grupo_b = total_grupo_b / len(grupo_b)

print(f"Grupo A → total: {total_grupo_a}, media: {media_grupo_a:.2f}")
print(f"Grupo B → total: {total_grupo_b}, media: {media_grupo_b:.2f}")


if media_grupo_a > media_grupo_b:
    print("Grupo A es el mejor.")
elif media_grupo_a < media_grupo_b:
    print("Grupo B es el mejor.")
else:
    print("Los dos grupos son iguales.")
```
